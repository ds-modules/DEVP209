# Histograms

A histogram depicts the frequency with which we observe the values on the x-axis in our DataFrame.

Suppose we are interested in the distribution of age in our DataFrame. We will want to plot the frequency with which different ages occur.

In practice, what this means is the following. We divide the data into equal-interval bins based on age. Then within each bin, count the number of observations present in the DataFrame whose age fall into that bin. Repeat for all bins. Then plot the result such that the height of each bar corresponds to the number of observations in each bin. 

```python
# First, let's pick the size of the bins we want to use 
# To start, let's use a bin size of 5 years 
bin_size = 5
```

```python
# Note that the maximum age in our DataFrame is 91
# Try to verify this
print(arrests['age'].max())
```

```python
# Now let's generate a set of number with the edges of the bins we want to use

# We can use the np.linspace function to do this
# The np.linspace function takes three arguments: the start value, the end value, and the number of values to generate

# We want to start at 0, end at 100, and generate (100/bin_size)+1 values (the +1 is because we want to include the endpoint)
# So we can use the following line of code to generate the bin edges

bins = np.linspace(0,100,int((100/bin_size))+1)

print(bins)
```

```python
# Recall that the plt function is a function imported from the matplotlib module (which we imported as plt)
# The plt.hist function takes two main arguments: the data to plot and the bin edges to use
# We want to plot the age column of the arrests DataFrame, and we want to use the bins we just generated

# Plot and label
plt.hist(arrests['age'], bins=bins)
plt.show()
```

```python
# Let's make this figure look nicer by adding a title and axis labels
plt.hist(arrests['age'], bins=bins)
plt.title('Histogram of Arrests by Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

# Scatterplots

Scatterplots are useful for visualizing the relationship between two variables. 

```python
# Let's load a new dataframe for this
# Note: if the input file uses tab spacing instead of comma separators, we need to specify the delimiter as '\t'
arrests_state_country = pd.read_csv('ArrestsStateCountry.csv', sep='\t')
arrests_state_country.head()
```

```python
# Do arrests per day for Trump and arrests per day for Biden correlate?
# Let's plot both against each other

# Create a scatterplot to visualize the relationship
# all we have to do here is call plt.scatter which takes two arguments (inputs)
# the first argument is the column of the dataframe corresponding to what we want as the x-axis variable
# the second argument is the column of the dataframe corresponding to what we want as the y-axis variable
plt.scatter(arrests_state_country['arrests_perdaybiden'], 
           arrests_state_country['arrests_perdaytrump'])

# we can also add titles as follows
plt.ylabel('Arrests per Day (Trump Administration)')
plt.xlabel('Arrests per Day (Biden Administration)')
plt.title('Arrests per Day Trump vs Biden by State-Country Combination')
plt.show()
```

```python
# Let's make the plot a little more interpretable by adding a grid and a 45 degree line 
plt.scatter(arrests_state_country['arrests_perdaybiden'], 
           arrests_state_country['arrests_perdaytrump'])
plt.xlabel('Arrests per Day (Biden Administration)')
plt.ylabel('Arrests per Day (Trump Administration)')
plt.title('Arrests per Day Trump vs Biden by State-Country Combination')

# adds a grid
plt.grid(True, alpha=0.2) #the alpha parameter controls transparent the grid is (0 is invisible, 1 is solid)

# add a 45 degree line to the plot
# first, we find the maximum value of arrests per day for either trump of biden
max_val = max(arrests_state_country['arrests_perdaybiden'].max(), 
              arrests_state_country['arrests_perdaytrump'].max())

# then we generate the 45 degree line
# the first argument [0, max_val] gives the x-axis values where we want to plot the line
# the second argument [0, max_val] gives the corresponding y-axis values
# 'r--' specifies a red dashed line
# alpha again controls the transparency
# finally label='45 degree line (y=x)' adds a label to the plot
plt.plot([0, max_val], [0, max_val], 'r--', alpha=0.5, label='45 degree line (y=x)')
plt.legend()
plt.tight_layout()
plt.show()
```

# OLS (Ordinary Least Squares) Regression

We can generate a best-fit line through our data using OLS regression. This statistical method finds the line that minimizes the sum of squared differences between the observed values and the predicted values.

Let's run a simple linear regression to see if there's a relationship between arrests per day under Trump and under Biden.

```python
# First, prepare the data for OLS regression

# Remove any rows with missing values
clean_data = arrests_state_country[['arrests_perdaybiden', 'arrests_perdaytrump']].dropna()
```

```python
# X variable (independent): arrests per day under Biden
# Y variable (dependent): arrests per day under Trump
X = clean_data['arrests_perdaybiden']
y = clean_data['arrests_perdaytrump']

# Add a constant term (intercept) to the model
X_with_constant = sm.add_constant(X)

# Fit the OLS model with the OLS function from the statsmodels (sm) module)
# the first argument is the outcome
# the second argument are the predictors (indepenent variables or "features")
# the fit() function actually fits the model 
model = sm.OLS(y, X_with_constant).fit()

# Print the summary table using the summary function
print(model.summary())
```

# Plotting the OLS fit 

Now let's plot the OLS fit and overlay it on the scatterplot

```python
# Create scatterplot with OLS best fit line
# Plot the original data points
plt.scatter(clean_data['arrests_perdaybiden'], 
           clean_data['arrests_perdaytrump'], 
           alpha=0.6, label='Data points')

# Create a range of x values for the best fit line
x_range = np.linspace(clean_data['arrests_perdaybiden'].min(), 
                      clean_data['arrests_perdaybiden'].max(), 100)

# Add constant term to x_range for prediction
x_range_with_const = sm.add_constant(x_range)

# Generate predictions using the fitted model
y_pred = model.predict(x_range_with_const)

# Plot the OLS best fit line
plt.plot(x_range, y_pred, 'r-', linewidth=2, label='OLS best fit line')

# Add labels and formatting
plt.xlabel('Arrests per Day (Biden Administration)')
plt.ylabel('Arrests per Day (Trump Administration)')
plt.title('Arrests per Day: Trump vs Biden with OLS Best Fit Line')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
