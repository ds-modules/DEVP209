```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set style for better looking plots
np.random.seed(42)  # For reproducible results

# Generate dummy data
# Dataset 1: Normal distribution (height data)
heights = np.random.normal(170, 10, 1000)  # mean=170cm, std=10cm

# Dataset 2: Skewed distribution (income data)
incomes = np.random.exponential(50000, 1000)  # exponential distribution

# Dataset 3: Bimodal distribution (test scores)
scores1 = np.random.normal(65, 8, 500)   # struggling students
scores2 = np.random.normal(85, 6, 500)   # high-performing students
test_scores = np.concatenate([scores1, scores2])

print(f"Heights: mean={heights.mean():.1f}, std={heights.std():.1f}")
print(f"Incomes: mean={incomes.mean():.0f}, median={np.median(incomes):.0f}")
print(f"Test scores: mean={test_scores.mean():.1f}, std={test_scores.std():.1f}")
```

```python
# 1. Basic Histogram
plt.hist(incomes, bins=50, alpha=0.6, color='lightgreen', edgecolor='black')    # first argument is a numpy array, bins gives number of bars to plot, other options for display
plt.title('Income Distribution (Skewed)')
plt.xlabel('Income ($)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.hist(test_scores, bins=30, alpha=0.7, color='salmon', edgecolor='black')
plt.title('Test Scores (Bimodal)')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 2. Scatterplots with dummy data - Positive Correlation and Non-linear Relationship
plt.figure(figsize=(12, 5))

# Generate additional variables for relationships
ages = np.random.uniform(18, 65, 1000)  # Random ages 18-65
experience = ages - 18 + np.random.normal(0, 3, 1000)  # Work experience roughly based on age
experience = np.clip(experience, 0, 50)  # Keep reasonable bounds

weight = 2.3 * heights - 220 + np.random.normal(0, 15, 1000)  # Weight roughly related to height
```

```python
# Scatterplot 1: Height vs Weight (positive correlation)
plt.subplot(1, 3, 1)
plt.scatter(heights, weight, alpha=0.6, color='blue', s=30)
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.title('Height vs Weight\n(Positive Correlation)')
plt.grid(True, alpha=0.3)

# Print correlation coefficients
print("Correlation Coefficients:")
print(f"Height vs Weight: {np.corrcoef(heights, weight)[0,1]:.3f}")
```
