### Kernel density plots 

Try generating a plot of the estimated pdf of age in the arrests data. As you learned in class, this is a plot of the distribution of the variable where the heights of the plot indicate the relative "likelihood" in which you would observe that value. It is *not* a probability. There are different ways you can estimate a density. The most common and flexible way to do it is via a method called "kernel density estimation" (KDE). The KDE procedure essentially computes the height of the density function at a certain value by using a weighted-average of data points located near that value. The method also ensures that the area under the curve of the density plot equals one (like a proper pdf should). You can read more about the mathematical details here: https://towardsdatascience.com/kernel-density-estimation-explained-step-by-step-7cc5b5bc4517/

We can use the seaborn module to apply KDE to our data. The following cell is an example. Make sure you understand each step. 

```python
# Create the KDE plot using seaborn
# Note that the syntax here is similar to the syntax you've already used for histograms and scatterplots
sns.kdeplot(data=arrests, x='age', fill=True, alpha=0.6) 

# Add labels and title
plt.xlabel('Age')
plt.ylabel('Density')
plt.title('Kernel Density Estimation of Age Distribution in Arrests Data')
plt.grid(False)
plt.show()
```

# Simulating random variables

In lecture, we learned about common distributions for random variables. We can use Python to visualize probability distributions. 

For example, below we plot the standard normal distribution. Reminder: the normal distribution is parameterized by the *mean* and *variance*. The standard normal has mean equal to zero and variance equal to one (which also means the standard deviation is equal to one). 

```python
# plot the standard normal pdf (just run this cell)
x = np.linspace(-4, 4, 1000)
pdf = (1/np.sqrt(2 * np.pi)) * np.exp(-0.5 * x**2) # this is the mathematical expression for the pdf of a standard normal
plt.plot(x, pdf)                                   # make sure you understand why x and pdf are the arguments in this function
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Standard Normal Distribution PDF')
plt.grid(False)  
plt.show()
```

We've plotted the pdf of a standard normal above. What if we have a sample that has been generated according to this distrbution? 

We can use Python to generate a samples with values drawn from a standard normal. (Just run the cell below.)

```python
# simulate 1000 realizations of a standard normal 
# the loc argument is the mean
# scale is the standard deviation
standard_normal_samples = np.random.normal(loc=0, scale=1, size=1000)

# plot the estimated density of the sample
sns.kdeplot(standard_normal_samples, fill=True, alpha=0.6)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Density Estimate of Standard Normal Samples')
plt.grid(True, alpha=0.3)
plt.show()
```

The empirical density looks like the pdf of the random variable it was drawn from. 

By plotting your data you can get a sense of how the data in our sample could have been generated.

### The fit appears to be much better. Why? 

As we know, OLS computes the line of best fit by minimizing the sum of squared residuals. But we've just shown that the fitted values actually form a curve and not a straight line. 

The key is that we've included $X$ and $X^2$ into the equation. This lets the effect of $X$ on $Y$ be *non-linear*.

To see this, suppose we have $Y = 2X + 3X^2$. What is the effect of a one unit change in $X$ on $Y$? Let's say we increase $X$ from 0 to 1. Then at $X = 0$, we have $Y=0$, and at $X=1$, we have $Y=5$. So $Y$ increased by 5 units.

Now let's suppose we increase $X$, again by one unit, but from $X=2$ to $X=3$. When $X=2$, we have $2(2) + 3(2)^2 = 16$. When $X=3$, we have $2(3) + 3(3)^2 = 33$. So the increase in $Y$ is 17, not 5.

The effect of a one unit change in $X$ depends on the value of $X$! Hence, the relationship is non-linear. We can use regression to model these non-linearities. An easy way to do so is to include higher-order terms as independent variables as we've done above.
