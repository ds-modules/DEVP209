## Using Python for confidence intervals

Let us practice using Python to calculate p-values and confidence intervals without using the complicated tables

Above, we imported from scipy.stats the norm library, which allows us to calculate values related to the *Normal Distribution*

Quick warm up:

Answers:

Q1: Cumulative density function. When we find the cumulative density, it tells us the total percentage of a distribution that occurs below a specific value. We are also able to express a range in the CDF function that tells us the total area under the probability curve in between two values (such as a confidence interval). A famous example is the total area below -1.96 and above 1.96 (WHY???? see below)

Q2: A confidence interval is the range that we expect a true value to lie within given a certain probability. If I have a 95% confidence interval, I expect a value that I measure for the mean to occur within a LOWER and UPPER BOUND to occur 95% of the time. We can still get outliers that are outside of this range

```python
# Please think of the answers before running this cell using what you know about confidence intervals

# Example 1: P(Z <= 1.96) #remember this is a one-sided confidence interval
p = norm.cdf(1.96)
print("P(Z <= 1.96) =", p)

# Example 2: P(Z > 1.96)
print("P(Z > 1.96) =", 1 - p)

# Example 3: z critical value for a 95% confidence level
z_crit = norm.ppf(0.975)
print("z critical value (95% CI) =", z_crit)
```

Okay so what we see above, as we remember from class is that a 95% confidence interval has a critical value of 1.96 (example 3)

A 95% confidence interval has an UPPER and LOWER BOUND, so the area outside of each of these bounds is 5%/2 = 2.5%, which is why Example 2 is 2.5%

Lastly, Example 1 has a value of 97.5% because we are not looking at the lower bound here of -1.96, so the value is 95% plus the 2.5% extra not included

### Plotting the normal distribution

```python
x = np.linspace(-4, 4, 400)
y = norm.pdf(x)

plt.plot(x, y, label='Standard Normal PDF')
plt.fill_between(x, 0, y, where=(x > -1.96) & (x < 1.96), color='skyblue', alpha=0.5, label='95% Area')
plt.title("Standard Normal Distribution")
plt.xlabel("Z")
plt.ylabel("Density")
plt.legend()
plt.show()
```

## Sample Mean

```python
# Suppose we have a population with mean 50 and standard deviation 10.
# We draw a random sample of n = 100.

population_mean = 50
population_sd = 10
n = 100

sample = np.random.normal(loc=population_mean, scale=population_sd, size=n) #We make a random normal distribution around our population data
sample_mean = np.mean(sample)
sample_var = np.var(sample, ddof=2)

print("Sample mean:", sample_mean)
print("Sample variance:", sample_var)
```

Note, in the above, ddof is what is subtracted from the divisor n in the variance formula. If you remember, for a sample, we divide by n-1

## Law of Large Numbers

```python
# Let's simulate many sample means to see how they behave
num_samples = 1000
n = 30
sample_means = []

for i in range(num_samples):
    x = np.random.normal(population_mean, population_sd, n)
    sample_means.append(np.mean(x))

plt.hist(sample_means, bins=30, density=True, alpha=0.7)
plt.axvline(population_mean, color='red', linestyle='dashed', label='True Mean')
plt.title("Sampling Distribution of the Sample Mean (n=30)")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.legend()
plt.show()

print("Mean of sample means:", np.mean(sample_means))
print("SD of sample means:", np.std(sample_means))
```

## Sample Variance

```python
from scipy.stats import t

# Example: Small sample, unknown sigma
x = np.array([45, 50, 47, 52, 44, 49, 46, 51])  # sample data
n = len(x)
xbar = np.mean(x)
s = np.std(x, ddof=1)  # sample standard deviation

confidence = 0.95
alpha = 1 - confidence

t_crit = t.ppf(1 - alpha/2, df=n-1)
margin = t_crit * s / np.sqrt(n)
ci = (xbar - margin, xbar + margin)

print("Sample mean:", round(xbar, 2))
print("Sample standard deviation:", round(s, 2))
print(f"{int(confidence*100)}% CI for mean (using t):", ci)
```

## Bonus: Why use Python?

So a lot of the above, you could have just done with a pencil, piece of paper, and normal distribution look up table. So why is Python important?

Here are some examples of why Python is valuable for data analysis, doing things that we otherwise couldn't

Example 1: 

Suppose test scores are normally distributed with
mean μ = 72 and standard deviation σ = 9.

What is the probability that a randomly chosen student scores above 85?

Method 1:
With paper, we could see Z = (X-mu)/sigma = (85-72)/9 = 1.44

P(X>85) = P(Z>1.44) = 1- 0.9251 (found in table) = 0.0749 meaning ~7.5% of students performed above 85 on this test

```python
# Define variables for Python method
mu = 72
sigma = 9
x_value = 85
```

```python
# Method 2: Use python and get answer immediately

# Use norm.cdf to get cumulative probability up to 85
p_below = norm.cdf(x_value, loc=mu, scale=sigma)
p_above = 1 - p_below

print("P(X > 85) =", p_above)
```

```python
# Method 3: Use python to convert to Z and find answer more slowly

z = (x_value - mu) / sigma
p_above_z = 1 - norm.cdf(z)
print("P(X > 85) using standard normal:", p_above_z)
```

### Example 2: What is the probability that a student's score lies between 60 and 90

```python
p_between = norm.cdf(90, mu, sigma) - norm.cdf(60, mu, sigma)
print("P(60 < X < 90) =", p_between)
```
