## Critical Values and Confidence Intervals

```python
# Let's set up a plot with the standard normal density 

# Create a range of x values
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)  # Standard normal probability density

plt.plot(x, y, 'k-', linewidth=2, label='Standard Normal')
```

Make a few observations about this plot. What is the mean/median/mode? 

```python
# Let's compute critical values for this distribution 

# Suppose we are interested in a 95% confidence level 
# store this value in conf_level 
conf_level = 0.95

# also, define alpha = 1 - conf_level 
# we would like to construct a confidence interval (CI) such that the probability the parameter is not contained in the CI is alpha (in repeated samples)
alpha = 1 - conf_level

# recall the norm.ppf returns the inverse of the CDF of the standard normal evaluated at its argument
# so we can obtain the 90th percentile of the normal distribution by using 
ninety_ptile = norm.ppf(.9)
print(ninety_ptile)

# We use the same logic to pick critical values
# We would like to pick critical values such that the probability falling above the larger critical value is 2.5% and falling below the smaller critical value is 2.5%
# Then, realizations of the standard normal should fall in the interval between the critival values with probability 100% - 2.5% - 2.5% = 95%
z_crit_upper = norm.ppf(1 - alpha/2)
z_crit_lower = norm.ppf(alpha/2)

# Let's print to see what the critical values are
print('larger critical value: ', z_crit_upper)
print('smaller critical value: ', z_crit_lower)
```

What do you notice about the critical values? They should be related to each other in a distinctive way. Why? 

Hint: look at the standard normal figure we've plotted above

Just run the following code block (no need to edit anything)

```python
plt.plot(x, y, 'k-', linewidth=2, label='Standard Normal')

# Shade the middle area (confidence interval)
mask = (x >= z_crit_lower) & (x <= z_crit_upper)
plt.fill_between(x[mask], y[mask], alpha=0.3, color='#3498db', label=f'{conf_level*100:.0f}% probability')

# Shade the rejection regions (tails)
tail_left = x < z_crit_lower
tail_right = x > z_crit_upper
plt.fill_between(x[tail_left], y[tail_left], alpha=0.2, color='red')
plt.fill_between(x[tail_right], y[tail_right], alpha=0.2, color='red', label=f'{alpha*100:.0f}% probability')

# Mark critical values
plt.axvline(z_crit_lower, color='#3498db', linestyle='--', linewidth=2, alpha=0.8)
plt.axvline(z_crit_upper, color='#3498db', linestyle='--', linewidth=2, alpha=0.8)

# Add text annotations
plt.text(0, max(y)*0.5, f'{conf_level*100:.0f}% of data\nfalls here', 
        ha='center', fontsize=12, bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
plt.text(z_crit_upper, max(y)*0.08, f'z = {z_crit_upper:.3f}', ha='center', fontsize=11, color='#3498db', weight='bold',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))
plt.text(z_crit_lower, max(y)*0.08, f'z = {z_crit_lower:.3f}', ha='center', fontsize=11, color='#3498db', weight='bold',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

plt.xlabel('Standard deviations from mean', fontsize=12)
plt.ylabel('Probability density', fontsize=12)
plt.title(f'{conf_level*100:.0f}% Confidence Level\nCritical value: ±{z_crit_upper:.3f}', fontsize=14, weight='bold')
plt.legend(loc='upper right', fontsize=11)
plt.grid(True, alpha=0.3)
plt.ylim([0, max(y)*1.1])

plt.tight_layout()
plt.show()
```

Critical values mark the boundary between 'typical' and 'unusual' results

For 95% confidence, we use critical values of -1.96 and 1.96

This also means 95% of sample estimates fall within ±1.96 standard deviations (because the x-axis is in standard deviation units)

### OLS inference

Let's put some of the tools in action. We will estimate a model via OLS and interpret its output

```python
# first, let's load the data (just run this cell)
arrests_state_country = pd.read_csv('ArrestsStateCountry.csv', sep='\t')
arrests_state_country.columns = arrests_state_country.columns.str.strip()
arrests_state_country.head()
```

```python
# Let's run a simple bivariate OLS regression using the arrests data

# Prepare the data
reg_cols = ['arrests_perdaybiden', 'arrests_perdaytrump']
reg_data = arrests_state_country[reg_cols].dropna().copy()

# Create the dependent variable (y) and independent variable (X)
y = reg_data['arrests_perdaybiden']
X = reg_data['arrests_perdaytrump']

# Add a constant term (intercept)
X = sm.add_constant(X)

# Fit the OLS model
model = sm.OLS(y, X).fit()

# Display the regression results
print(model.summary())
```

We've fit a model relating arrests_perdaybiden to arrests_perdaytrump.

```python
# Simulate data with 1000 observations
np.random.seed(42)  # For reproducibility

# Number of observations
n = 1000

# Generate the independent variable X
# Let's make it between 0 and 10
X_sim = np.random.uniform(0, 10, n)

# Generate the dependent variable Y with a linear relationship plus noise
# True model: Y = 2 + 3*X + error
true_intercept = 2
true_slope = 3
error = np.random.normal(0, 2, n)  # Normal errors with std dev = 2

Y_sim = true_intercept + true_slope * X_sim + error

# Now run OLS regression
X_reg = sm.add_constant(X_sim)
model_sim = sm.OLS(Y_sim, X_reg).fit()

print(model_sim.summary())
```
