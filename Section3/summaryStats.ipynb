Working with DataFrames:

We have two different ways to interact with DataFrames. This is a broad generalization, but let us say right now we can analyze them by removing/adding values and filtering. Or we can try to change them in more complicated ways such as comparing them to other DataFrames (for one example)

When we analyze a DataFrame, we are determining its attributes, which means we use DataFrame "methods". This is basically done by writing the DataFrame name, adding a "." and then writing the method you are interested in with any arguments.

For instance, if I want to see the top 10 rows, this is me analyzing the DataFrame. I know I can see the top X rows using the "head" method with an argument 10, as below

The way we filter is by using the "loc" which basically means return the DataFrame information located using a certain condition. For instance, let me select the Column 'Departed Date' where 'Apprehension State' is unknown and 'Birth Year' is after 2000.

Note the use of parentheses to contain my filtering statement

````python
arrests.loc[(pd.isnull(arrests['Departed Date']))&(arrests['Birth Year']>2000)]
````

This is a complicated expression. Let us start with

1. ```arrests```
A pandas DataFrame (like a spreadsheet with rows and columns).

2. ```arrests['Departed Date']```
Selects the column named Departed Date.

3. ```pd.isnull(arrests['Departed Date'])```
Checks which rows have missing values in Departed Date.
Returns True for missing, False otherwise.
This uses the imported function ```isnull``` from the Pandas library (which we aliased above to be called ```pd```)

4. ```arrests['Birth Year'] > 2000```
Checks which rows have a Birth Year greater than 2000.
Returns True or False for each row.

5. ```(... ) & (... )```
Combines the two conditions with a logical AND denoted by ```&```.
A row is True only if both conditions are True.

6. ```arrests.loc[...]```
Selects only the rows where the combined condition is True

````python
# For arrests, the data does not have age (or birth date) but it does have birth year, so bucket by that
#Find the current year
current_year = datetime.datetime.today().year #2025
arrests['age'] = current_year - arrests['Birth Year']
arrests['Apprehension Date'] = pd.to_datetime(arrests['Apprehension Date'],format='%m/%d/%Y %H:%M')
arrests['Apprehension Day'] = arrests['Apprehension Date'].dt.date
````

````python
# Try to look at the DataFrame here and understand what each column means
````

These are basic ways to find statistics using a dataframe series or numpy array

````python
# Ways to find Mean
mean_pandas = arrests['age'].mean()
mean_manual = sum(arrests['age']) / len(arrests['age'])
def find_mean_verbose(series):
    tempSum = 0
    count = len(series)
    for val in series:
        tempSum = tempSum + val
    return tempSum/count
mean_verbose = find_mean_verbose(arrests['age'])

print(f"Mean (Pandas): {mean_pandas}")
print(f"Mean (Manual): {mean_manual}")
print(f"Mean (Verbose): {mean_verbose}")
````

````python
# --- MEDIAN ---
median_pandas = arrests['age'].median()
median_manual = np.median(arrests['age'])

print(f"\nMedian (Pandas): {median_pandas}")
print(f"Median (Manual - NumPy): {median_manual}")
````

````python
# --- MODE ---
mode_pandas = arrests['age'].mode()
mode_manual = max(set(arrests['age']), key=list(arrests['age']).count)  # Only shows one mode

print(f"\nMode (Pandas): {mode_pandas.tolist()} (can show multiple modes)")
print(f"Mode (Manual): {mode_manual}")
````

````python
# --- VARIANCE ---
variance_pandas = arrests['age'].var()  # Sample variance (ddof=1)
variance_manual = np.var(arrests['age'], ddof=1)
def find_variance_verbose(series):
    count = len(series)
    mean = series.mean()
    runningSoS = 0
    for val in series:
        runningSoS = runningSoS + (val-mean)**2
    return runningSoS/(count-1)
variance_verbose = find_variance_verbose(arrests['age'])

print(f"\nVariance (Pandas): {variance_pandas:.2f}")
print(f"Variance (Manual - NumPy): {variance_manual:.2f}")
print(f"Variance (Verbose): {variance_verbose:.2f}")
````

### These are more complicated ways but allow some grouping by other values

````python
arrests.groupby('Apprehension State')['age'].agg(
    mean='mean',
    median='median',
    mode=lambda x: x.mode().iloc[0] if not x.mode().empty else None,
    count='count'
)
````

````python
arrests.groupby(['Apprehension State','Gender'])['age'].agg(
    mean='mean',
    median='median',
    mode=lambda x: x.mode().iloc[0] if not x.mode().empty else None,
    count='count'
)
````
